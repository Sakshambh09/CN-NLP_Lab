```python
# Cell 3: Toolkit-based TF Analysis and NER
import spacy
from collections import Counter
import pandas as pd

# Load the spaCy English model
nlp = spacy.load("en_core_web_sm")

# Read the input text file
with open("input_text.txt", "r", encoding="utf-8") as f:
    text_41 = f.read()

# Process text with spaCy
doc = nlp(text_41)

# Extract word tokens (ignoring punctuation, spaces, and optionally symbols)
words = [token.text.lower() for token in doc if not token.is_punct and not token.is_space]

# Calculate word frequencies
tf_counter = Counter(words)

# Convert to DataFrame and save to CSV
df_tf_toolkit = pd.DataFrame(tf_counter.items(), columns=["Term", "Frequency"])
df_tf_toolkit.sort_values(by="Frequency", ascending=False, inplace=True)
df_tf_toolkit.to_csv("term_frequency_toolkit.csv", index=False)
print("Saved toolkit Term Frequency to 'term_frequency_toolkit.csv'.\n")

# Display top 10 most frequent terms
print("Top 10 Most Frequent Terms (Toolkit):")
print(df_tf_toolkit.head(10).to_string(index=False))

# Named Entity Recognition (NER)
print("\n" + "="*50)
print("Named Entities Identified:")
ner_records = [{"Entity": ent.text.strip(), "Label": ent.label_} for ent in doc.ents]
df_ner = pd.DataFrame(ner_records)

# Display first 20 recognized entities
print(df_ner.drop_duplicates().head(20).to_string(index=False))
```

```python
# Cell 4: Native Python TF Analysis
import string
import csv

# Read the input text file
with open("input_text.txt", "r", encoding="utf-8") as f:
    text_42 = f.read()

# Preprocessing: convert to lowercase and remove punctuation
text_cleaned = text_42.lower()
for char in string.punctuation:
    text_cleaned = text_cleaned.replace(char, " ")

# Tokenize using whitespace
tokens = text_cleaned.split()

# Frequency calculation using a native Python dictionary
frequency_dict = {}
for word in tokens:
    if word in frequency_dict:
        frequency_dict[word] += 1
    else:
        frequency_dict[word] = 1

# Sort terms by frequency in descending order
sorted_terms = sorted(frequency_dict.items(), key=lambda item: item[1], reverse=True)

# Save output to CSV
csv_filename = "term_frequency_no_toolkit.csv"
with open(csv_filename, mode="w", newline="", encoding="utf-8") as csv_file:
    writer = csv.writer(csv_file)
    writer.writerow(["Term", "Frequency"])
    for term, freq in sorted_terms:
        writer.writerow([term, freq])

print(f"Frequency table successfully saved to '{csv_filename}'.\n")

# Display top 10 most frequent terms
print("Top 10 Most Frequent Terms (Without Toolkit):")
print(f"{'Term':<20} {'Frequency':<10}")
print("-" * 30)
for term, freq in sorted_terms[:10]:
    print(f"{term:<20} {freq:<10}")
```

```python
# Cell 5: Define Documents, Preprocess, and Tokenize
import math
import string

# Define the given documents
documents = {
    "Document 1": "Natural language processing is a field of artificial intelligence.",
    "Document 2": "Natural language processing helps computers understand human language.",
    "Document 3": "Machine learning is an important part of artificial intelligence."
}

# Preprocessing and Tokenization
tokenized_docs = {}
for doc_id, text in documents.items():
    cleaned_text = text.lower()
    for p in string.punctuation:
        cleaned_text = cleaned_text.replace(p, " ")
    tokenized_docs[doc_id] = cleaned_text.split()

# Display tokenization output
for doc_id, tokens in tokenized_docs.items():
    print(f"{doc_id} Tokens ({len(tokens)}): {tokens}")
```

```python
# Cell 6: Step-by-Step TF, DF, IDF, and TF-IDF Computation
# 1. Calculate Term Frequency (TF) for each document: count(t, d) / total_words(d)
tf_scores = {}
for doc_id, tokens in tokenized_docs.items():
    total_words = len(tokens)
    counts = {}
    for token in tokens:
        counts[token] = counts.get(token, 0) + 1
    tf_scores[doc_id] = {token: count / total_words for token, count in counts.items()}

# 2. Extract vocabulary and compute Document Frequency (DF)
vocab = sorted(list(set(word for tokens in tokenized_docs.values() for word in tokens)))
N = len(documents)

df_scores = {}
for term in vocab:
    df_scores[term] = sum(1 for tokens in tokenized_docs.values() if term in tokens)

# 3. Calculate Inverse Document Frequency (IDF): log(N / DF)
idf_scores = {}
for term in vocab:
    idf_scores[term] = math.log(N / df_scores[term])

# 4. Calculate TF-IDF: TF * IDF
tfidf_scores = {}
for doc_id in documents:
    tfidf_scores[doc_id] = {}
    for term in tf_scores[doc_id]:
        tfidf_scores[doc_id][term] = tf_scores[doc_id][term] * idf_scores[term]

# Display Document Frequency (DF) and Inverse Document Frequency (IDF)
print(f"{'Term':<16} {'DF':<6} {'IDF':<10}")
print("-" * 34)
for term in vocab:
    print(f"{term:<16} {df_scores[term]:<6} {idf_scores[term]:<10.4f}")
```

```python
# Cell 7: Display Final Scores and Top 10 Terms Per Document
for doc_id in documents:
    print(f"\n{'='*55}\nResults for {doc_id}\n{'='*55}")
    print(f"{'Term':<16} {'TF':<10} {'DF':<6} {'IDF':<10} {'TF-IDF':<10}")
    print("-" * 55)
    for term in tokenized_docs[doc_id]:
        # Print each unique term's calculated values
        if term in tf_scores[doc_id]:
            tf = tf_scores[doc_id][term]
            df = df_scores[term]
            idf = idf_scores[term]
            tfidf = tfidf_scores[doc_id][term]
            print(f"{term:<16} {tf:<10.4f} {df:<6} {idf:<10.4f} {tfidf:<10.4f}")

    # Top 10 terms with the highest TF-IDF score
    sorted_tfidf = sorted(tfidf_scores[doc_id].items(), key=lambda item: item[1], reverse=True)
    print(f"\nTop Terms (Highest TF-IDF) in {doc_id}:")
    for rank, (term, score) in enumerate(sorted_tfidf[:10], start=1):
        print(f"{rank}. {term:<15} : {score:.4f}")
```
