**Experiment 2.1: Basic Text Preprocessing**

```python
import re
import string

# Load input file
file_path = "2.1_text_data.txt"
with open(file_path, "r", encoding="utf-8") as file:
    text = file.read()

# 1. Lowercase conversion & count uppercase letters
uppercase_count = sum(1 for char in text if char.isupper())
lowercase_text = text.lower()

# 2. Punctuation removal & count distinct punctuation types
found_punctuations = set(char for char in text if char in string.punctuation)
punctuation_types_count = len(found_punctuations)
no_punctuation_text = lowercase_text.translate(str.maketrans('', '', string.punctuation))

# 3. Numbers removal & count total digits/numbers present
digits_count = len(re.findall(r'\d', text))
no_numbers_text = re.sub(r'\d+', '', no_punctuation_text)

# 4. Extra whitespace removal & count extra spaces removed
cleaned_text = re.sub(r'\s+', ' ', no_numbers_text).strip()
extra_whitespace_count = len(no_numbers_text) - len(cleaned_text)

# Display Analysis Metrics
print(f"Uppercase Characters Count: {uppercase_count}")
print(f"Distinct Punctuation Types Found: {punctuation_types_count} -> ({', '.join(sorted(found_punctuations))})")
print(f"Total Numbers/Digits Found: {digits_count}")
print(f"Extra Whitespace Characters Removed: {extra_whitespace_count}")

print("\n--- CLEANED TEXT ---")
print(cleaned_text)
```

**Experiment 2.2: Word and Sentence Tokenization**

```python
import nltk
from nltk.tokenize import sent_tokenize, word_tokenize

nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)

# Load input file
file_path = "2.2_tokenization_data.txt"
with open(file_path, "r", encoding="utf-8") as file:
    text_data = file.read()

# Tokenization
sentences_nltk = sent_tokenize(text_data)
words_nltk = word_tokenize(text_data)

print("--- SENTENCE TOKENIZATION (NLTK) ---")
for i, sent in enumerate(sentences_nltk, 1):
    print(f"{i}. {sent}")

print("\n--- WORD TOKENIZATION (NLTK) ---")
print(words_nltk)
```

```python
import spacy

nlp = spacy.load("en_core_web_sm")

# Load input file
file_path = "2.2_tokenization_data.txt"
with open(file_path, "r", encoding="utf-8") as file:
    text_data = file.read()

doc = nlp(text_data)

# Tokenization
sentences_spacy = [sent.text.strip() for sent in doc.sents]
words_spacy = [token.text for token in doc if not token.is_space]

print("--- SENTENCE TOKENIZATION (spaCy) ---")
for i, sent in enumerate(sentences_spacy, 1):
    print(f"{i}. {sent}")

print("\n--- WORD TOKENIZATION (spaCy) ---")
print(words_spacy)
```

```python
import re

# Load input file
file_path = "2.2_tokenization_data.txt"
with open(file_path, "r", encoding="utf-8") as file:
    text_data = file.read()

# Sentence tokenization using regular expressions
sentences_custom = [s.strip() for s in re.split(r'(?<=[.!?])\s+', text_data) if s.strip()]

# Word tokenization using regular expressions
words_custom = re.findall(r'\b\w+\b|[^\w\s]', text_data)

print("--- SENTENCE TOKENIZATION (Custom/Python Standard) ---")
for i, sent in enumerate(sentences_custom, 1):
    print(f"{i}. {sent}")

print("\n--- WORD TOKENIZATION (Custom/Python Standard) ---")
print(words_custom)
```

**Experiment 2.3: Stop Words Removal**

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

nltk.download('stopwords', quiet=True)
nltk.download('punkt', quiet=True)

# Load input file
file_path = "2.3_clean_data.txt"
with open(file_path, "r", encoding="utf-8") as file:
    text_data = file.read()

# 1. Tokenize original text
original_tokens = word_tokenize(text_data)

# 2. Get stop words set
stop_words_set = set(stopwords.words('english'))

# 3. Categorize tokens
detected_stop_words = [token for token in original_tokens if token.lower() in stop_words_set]
filtered_tokens = [token for token in original_tokens if token.lower() not in stop_words_set]

# Display Results
print("--- ORIGINAL TOKENS ---")
print(original_tokens)

print(f"\n--- STOP WORDS SEPARATELY (Total Detected: {len(detected_stop_words)}) ---")
print(detected_stop_words)

print("\n--- FILTERED TOKENS (Stop Words Removed) ---")
print(filtered_tokens)
```
